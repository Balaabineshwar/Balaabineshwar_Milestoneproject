```python
import pandas as pd

print("=== 1. LOADING ALL MANDATORY OLIST DATASETS ===")
customers = pd.read_csv('olist_customers_dataset.csv')
orders = pd.read_csv('olist_orders_dataset.csv')
order_items = pd.read_csv('olist_order_items_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
payments = pd.read_csv('olist_order_payments_dataset.csv')
reviews = pd.read_csv('olist_order_reviews_dataset.csv')
sellers = pd.read_csv('olist_sellers_dataset.csv')
print("All 7 datasets loaded successfully!\n")
```

```python
print("=== 2. VERIFYING DATASET DIMENSIONS (.shape) ===")
print("Customers shape:", customers.shape)
print("Orders shape:", orders.shape)
print("Order Items shape:", order_items.shape)
print("Products shape:", products.shape)
print("Payments shape:", payments.shape)
print("Reviews shape:", reviews.shape)
print("Sellers shape:", sellers.shape)
```

```python
# Checking duplicate rows in all 7 datasets
print("=== 3.DUPLICATES COUNT ===")
print("Customers:", customers.duplicated().sum())
print("Orders:", orders.duplicated().sum())
print("Order Items:", order_items.duplicated().sum())
print("Products:", products.duplicated().sum())
print("Payments:", payments.duplicated().sum())
print("Reviews:", reviews.duplicated().sum())
print("Sellers:", sellers.duplicated().sum())
```

```python
# Checking missing values in all 7 datasets
print("=== 4.MISSING VALUES ===")
print("Customers:\n", customers.isnull().sum())
print("\nOrders:\n", orders.isnull().sum())
print("\nOrder Items:\n", order_items.isnull().sum())
print("\nProducts:\n", products.isnull().sum())
print("\nPayments:\n", payments.isnull().sum())
print("\nReviews:\n", reviews.isnull().sum())
print("\nSellers:\n", sellers.isnull().sum())
```
